```python
#!/usr/bin/env python3
"""
Plot energy convergence for QAOA and HEA-VQE
from benchmark_pl.py output (e.g., maxcut12.csv).
"""

import pandas as pd
import matplotlib.pyplot as plt

```

```python
# === Load CSV file ===
csv_file = "maxcut12.csv"   # path to your file
df = pd.read_csv(csv_file)

# === Separate data for QAOA and HEA ===
qaoa_df = df[df["algo"].str.contains("QAOA")].copy()
hea_df = df[df["algo"].str.contains("HEA")].copy()
```

```python
# === Plot setup ===
plt.figure(figsize=(7, 4))
plt.plot(qaoa_df["step"], qaoa_df["energy"],
         label="QAOA (p=2)", linewidth=2)

# === Formatting ===
plt.xlabel("Optimization Step", fontsize=11)
plt.ylabel("Energy Expectation Value", fontsize=11)
plt.title("Energy Convergence on 12-Qubit 3-Regular Max-Cut", fontsize=12)
plt.grid(True, linestyle="--", alpha=0.6)
plt.legend(fontsize=10)
plt.tight_layout()

# === Save or show ===
plt.savefig("QAOA.png", bbox_inches="tight")
plt.show()
```

```python

# === Plot setup ===
plt.figure(figsize=(7, 4))
plt.plot(hea_df["step"], hea_df["energy"],
         label="HEA-VQE (d=3)", linewidth=2)

# === Formatting ===
plt.xlabel("Optimization Step", fontsize=11)
plt.ylabel("Energy Expectation Value", fontsize=11)
plt.title("Energy Convergence on 12-Qubit 3-Regular Max-Cut", fontsize=12)
plt.grid(True, linestyle="--", alpha=0.6)
plt.legend(fontsize=10)
plt.tight_layout()

# === Save or show ===
plt.savefig("HEA.png", bbox_inches="tight")
plt.show()
```
